## TensorBoard Visualization & Model Export

### TensorBoard Monitoring

**Purpose**: Track training and evaluation metrics to:
1. Monitor loss convergence
2. Detect overfitting (divergence between train/eval loss)
3. Compare learning dynamics between SSD and Faster R-CNN

**Key Metrics Visualized**:
- `Loss/total_loss`: Combined classification + localization loss
- `Loss/classification_loss`: Class prediction error
- `Loss/localization_loss`: Bounding box regression error
- `DetectionBoxes_Precision/mAP`: COCO-style mean average precision

Separate TensorBoard instances launched on different ports for simultaneous model comparison.

````python
import os
import subprocess

ssd_log_dir = r"C:/Users/MSC1/Desktop/models/ssd_resnet50_v1_fpn_640x640_coco17_tpu-8/training_output"

def launch_tensorboard(log_dir, port=6006):
    print(f"Launching TensorBoard at http://localhost:{port}/")
    subprocess.run(["tensorboard", f"--logdir={log_dir}", f"--port={port}"])

launch_tensorboard(ssd_log_dir, port=6006)
````

## Model Export (Freezing)

**Purpose**: Convert trained checkpoints to deployment-ready SavedModel format for inference.

**Export Process**:
1. Load final checkpoint from `training_output/`
2. Freeze graph weights (no longer trainable)
3. Export as TensorFlow SavedModel with `image_tensor` input signature

**Output Structure**:
```
exported_model/
├── saved_model/
│   ├── saved_model.pb          # Frozen graph definition
│   └── variables/              # Trained weights
└── pipeline.config             # Model configuration
```

The exported models can be loaded directly via `tf.saved_model.load()` for inference.

````python
PIPELINE_CONFIG_PATH = r"C:/Users/MSC1/Desktop/models/ssd_resnet50_v1_fpn_640x640_coco17_tpu-8/pipeline.config"
TRAINED_MODEL_DIR = r"C:/Users/MSC1/Desktop/models/ssd_resnet50_v1_fpn_640x640_coco17_tpu-8/training_output"
OUTPUT_DIR = r"C:/Users/MSC1/Desktop/models/ssd_resnet50_v1_fpn_640x640_coco17_tpu-8/exported_model"
EXPORT_SCRIPT_PATH = r"C:/Users/MSC1/Desktop/Tensorflow-Object-Detection-API/models/research/object_detection/exporter_main_v2.py"

os.makedirs(OUTPUT_DIR, exist_ok=True)

export_command = [
    "python", EXPORT_SCRIPT_PATH,
    f"--input_type=image_tensor",
    f"--pipeline_config_path={PIPELINE_CONFIG_PATH}",
    f"--trained_checkpoint_dir={TRAINED_MODEL_DIR}",
    f"--output_directory={OUTPUT_DIR}"
]

print("Freezing the trained model...")
process = subprocess.Popen(export_command, stdout=subprocess.PIPE, stderr=subprocess.PIPE, text=True)

stdout, stderr = process.communicate()

print("=== STDOUT ===")
print(stdout)
print("=== STDERR ===")
print(stderr)

if process.returncode == 0:
    print(f"Model successfully exported to {OUTPUT_DIR}")
else:
    print(f"Model freezing failed with return code {process.returncode}. Check the logs above for details.")
````
