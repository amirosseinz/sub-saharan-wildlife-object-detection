## Model Evaluation

**Objective**: Assess trained model performance using COCO-style evaluation metrics to determine detection accuracy and localization quality.

**Key Metrics**:
1. **Mean Average Precision (mAP)**: Primary detection performance metric
   - AP @ IoU=0.50:0.95 (averaged over multiple IoU thresholds)
   - AP @ IoU=0.50 (standard detection threshold)
   - AP @ IoU=0.75 (strict localization threshold)

2. **Average Recall (AR)**: Detection completeness metric
   - AR @ maxDets=10: Recall with limited detections
   - AR @ maxDets=100: Recall with more detection allowance

3. **IoU (Intersection over Union)**: Localization quality metric
   - Measures overlap between predicted and ground truth bounding boxes
   - IoU=0.50 is standard; IoU=0.75+ indicates precise localization

**Monitoring for Overfitting**:
- Compare training loss vs evaluation loss trends
- Check if evaluation metrics plateau or degrade while training loss continues decreasing

## Comparative Analysis & Conclusions

### Model Comparison Summary

| Metric | SSD ResNet50 FPN | Faster R-CNN ResNet101 | Winner |
|--------|------------------|------------------------|--------|
| AP @ IoU=0.50:0.95 | **0.839** | 0.784 | SSD |
| AP @ IoU=0.50 | 0.910 | **0.912** | Faster R-CNN (marginal) |
| AP @ IoU=0.75 | **0.868** | 0.849 | SSD |
| AR (maxDets=10) | **0.860** | 0.820 | SSD |
| Final Loss | 0.252 | **0.076** | Faster R-CNN |
| Training Time | **~6 hours** | ~14 hours | SSD |
| Inference Speed | **46ms** | ~100ms | SSD |

### Key Findings

1. **Detection Performance**: Both models achieve excellent detection rates (AP@0.50 > 0.91), suitable for wildlife monitoring applications.

2. **Localization Quality**: SSD demonstrates better IoU metrics at stricter thresholds (AP@0.75), suggesting tighter bounding box predictions despite its single-stage architecture.

3. **Speed/Accuracy Trade-off**: SSD provides comparable accuracy with significantly faster training and inference, making it more practical for deployment on resource-constrained edge devices (e.g., camera traps).

4. **Overfitting**: Neither model shows signs of overfitting—evaluation metrics remain strong relative to training loss trends.

### Recommendations

- **Real-time deployment**: Use SSD ResNet50 FPN for faster inference
- **Maximum accuracy**: Faster R-CNN may be preferred when inference speed is not critical
- **Future work**: Test higher input resolutions (1024×1024) to improve small object detection, particularly for distant animals in camera trap imagery

### Limitations

1. Class balance was computed at the file level rather than annotation level in the original analysis (addressed in preprocessing corrections)
2. Input resolution (640×640) may undersample fine details from native camera trap resolutions
3. Additional IoU threshold analysis could provide more granular localization insights
